```python
#from delta format to csv
from pyspark.sql.functions import col, concat_ws
from pyspark.sql.types import ArrayType

df = spark.read.format("delta").load('/Volumes/hr_analytics/hr_schema/cleansed_hr_data/deltasave')

df_fixed = df.select([
    concat_ws(",", col(c)).alias(c) if isinstance(df.schema[c].dataType, ArrayType)
    else col(c)
    for c in df.columns
])

df_fixed.write.mode("overwrite").option("header","true") \
    .csv('/Volumes/hr_analytics/hr_schema/cleansed_hr_data/output_csv')
```

Why do we use .load() in PySpark?

👉 .load() is used to actually read the data from a path after specifying the format.

✅ Basic Concept
spark.read.format("delta").load("path")

👉 Think like this:

Part	Meaning
spark.read	Start reading
.format("delta")	Tell Spark file type
.load("path")	Load data from location
🧠 Simple Analogy

👉 Imagine:

.format() = "What type of file?"
.load() = "From where?"

Why I used different styles there?
# Read
spark.read.json("path_json")
spark.read.csv("path_csv", header=True)
spark.read.format("delta").load("path_delta")

👉 I used shortcut methods for JSON & CSV
👉 And generic method for Delta

✅ Correct Understanding
🔹 1. JSON & CSV → Have shortcuts

These are equivalent:

# Shortcut
spark.read.json("path_json")

# Generic
spark.read.format("json").load("path_json")
# Shortcut
spark.read.csv("path_csv", header=True)

# Generic
spark.read.format("csv").option("header","true").load("path_csv")

👉 Both are SAME internally ✅

🔹 2. Delta → No shortcut exists
spark.read.format("delta").load("path_delta")   ✅

👉 This is the ONLY way

❌ This does NOT exist:

spark.read.delta("path")   # invalid
🔥 Why I mixed styles?

👉 Just for simplicity:

JSON/CSV → shorter syntax
Delta → only available syntax

But in real projects:
👉 It’s better to be consistent

✅ Best Practice (Recommended)

Use one style everywhere 👇

spark.read.format("json").load("path_json")
spark.read.format("csv").option("header","true").load("path_csv")
spark.read.format("delta").load("path_delta")

👉 Clean + professional + interview-friendly ✅

🧠 Interview Answer

"Spark provides shortcut methods like read.csv() and read.json(), but internally they use format() and load(). Delta does not have a shortcut method, so we always use format('delta').load()."

⚡ Final Rule

👉 Two valid approaches:

✔️ Shortcut (easy)
spark.read.csv()
spark.read.json()
✔️ Generic (recommended)
spark.read.format().load()
🚀 Pro Tip (Important for you)

In interviews and real projects:
👉 Prefer this 👇
