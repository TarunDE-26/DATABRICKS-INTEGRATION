```python
%sql
CREATE CATALOG IF NOT EXISTS tarun_catalog;
CREATE SCHEMA IF NOT EXISTS workspace.sales_schema;

USE workspace.sales_schema;

CREATE TABLE IF NOT EXISTS sales_table (
    id INT,
    product STRING,
    price INT
)
USING DELTA;

INSERT INTO sales_table VALUES
(1, 'mobile', 10000),
(2, 'laptop', 50000);

SELECT * FROM sales_table;
```
