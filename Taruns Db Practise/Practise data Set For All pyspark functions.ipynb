```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DateType, BooleanType

schema = StructType([
    StructField("emp_id", IntegerType(), True),
    StructField("name", StringType(), True),
    StructField("city", StringType(), True),
    StructField("start_date", StringType(), True),  # Use StringType if dates are in string format
    StructField("end_date", StringType(), True),    # Use StringType for None values
    StructField("current_flag", BooleanType(), True)
])

old_data = spark.createDataFrame(
    [
        (1, "Tarun", "Hyderabad", "2023-01-01", None, True),
        (2, "Rahul", "Bangalore", "2023-01-01", None, True),
        (3, "Arjun", "Chennai", "2023-01-01", None, True),
    ],
    schema
)

display(old_data)
```

```python
new_data = spark.createDataFrame(
    [
        (1, "Tarun", "Hyderabad"),   # no change
        (2, "Rahul", "Mumbai"),      # changed city
        (4, "Kiran", "Pune")         # new employee
    ],
    ["emp_id", "name", "city"]
)

display(new_data)
```

```python
from delta.tables import DeltaTable

# Convert old_data into Delta table (if not already)
old_data.write.format("delta").mode("overwrite").save("/tmp/scd1_table")

scd1_table = DeltaTable.forPath(spark, "/tmp/scd1_table")

(scd1_table.alias("t")
 .merge(
      new_data.alias("s"),
      "t.emp_id = s.emp_id"
 )
 .whenMatchedUpdateAll()
 .whenNotMatchedInsertAll()
 .execute()
)
```
